<h3><b>המרת הנתונים לערכים נומריים (Data Preprocessing)</b></h3>

<p>בשלב זה של הכנת הנתונים, נבצע שתי פעולות מרכזיות כדי להתאים את הדאטה למודלים של למידת מכונה, אשר דורשים קלט מספרי בלבד:</p>

<ol>
  <li><b>המרת משתנים רציפים לקטגוריות (Binning / Discretization):</b><br>
  ניקח את עמודת הגיל (<code>age</code>), שהיא משתנה רציף, ונחלק אותה ל-3 טווחים שונים: צעיר (1), בוגר (2) ומבוגר (3). חלוקה זו מסייעת למודל להכליל תבניות ולהיות פחות רגיש להבדלים קטנים בגיל (למשל, בין גיל 22 ל-23).</li>
<br>
  <li><b>המרת משתנים קטגוריאליים למספרים (Label Encoding):</b><br>
  נמיר את העמודות שמכילות טקסט (Strings) כמו אזור (<code>region</code>) וסגנון פתיחה (<code>primary_opening_style</code>), לערכים מספריים באמצעות מיפוי (Mapping). כל מחרוזת ייחודית תקבל ערך מספרי קבוע.</li>
</ol>
</div>

```python
# 1. המרת גילאים (Age) לקטגוריות נומריות
# חלוקה לקבוצות גיל
bins = [0, 25, 55, 120]
labels = [1, 2, 3] # 1: צעיר, 2: בוגר, 3: מבוגר
df['age_group_numeric'] = pd.cut(df['age'], bins=bins, labels=labels)

# 2. המרת מחרוזות (Strings) למספרים בעזרת מילונים
region_mapping = {
    'Asia': 1,
    'Americas': 2,
    'East Europe & Russia': 3,
    'Western Europe': 4,
    'Rest of World': 5
}
df['region_numeric'] = df['region'].map(region_mapping)

time_control_mapping = {
    'Bullet': 1,
    'Blitz': 2,
    'Rapid': 3,
    'Classical': 4
}
df['preferred_time_control_numeric'] = df['preferred_time_control'].map(time_control_mapping)

opening_mapping = {
    'Positional': 1,
    'Defensive': 2,
    'Aggressive': 3
}
df['primary_opening_style_numeric'] = df['primary_opening_style'].map(opening_mapping)

premium_tier_mapping = {
    'Free': 1,
    'Pro': 2,
    'Elite': 3
}
df['chess_events_il_premium_tier_numeric'] = df['chess_events_il_premium_tier'].map(premium_tier_mapping)

# הדפסת הצצה לנתונים החדשים
df[['age', 'age_group_numeric', 'region', 'region_numeric']].head()
```
